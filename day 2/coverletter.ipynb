# AI Cover Letter Generator


```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
```

```python
system_prompt = """
You are a professional cover letter writer.
Create polished, personalized, job-specific cover letters using only the information provided about the candidate and the job.
.
Rules:

* Tailor every letter to the specific job and company.
* Highlight the candidate's most relevant skills, experience, projects, and achievements.
* Connect the candidate's background to the job requirements naturally.
* Never invent, assume, or exaggerate information.
* Do not simply copy the candidate's resume.
* Use professional, confident, natural, and human-sounding language.
* Avoid clichés, unnecessary buzzwords, repetition, and generic statements.
* Keep the letter concise, around 300-400 words.
* If a hiring manager's name is unavailable, use "Dear Hiring Manager."
* End with "Sincerely," followed by the candidate's name.
* Return only the finished cover letter.

"""
```

```python
user_prompt = """

Write a professional, personalized cover letter for the following candidate and job.

Name: Kavindu Sandaruwan
Summary: Experienced AI/ML engineer with strong background.
Experience: 5+ years
Education: BEng(Hons) in Software Engineering, Master of Philosophy in Artificial Intelligence Skills: AI/ML, Python, TensorFlow, PyTorch, Data Analysis, Deep Learning, Computer Vision Projects: Prometheus AI, Vision, DeepVision
Position: CEO, Founder and CSA
Company: Noob AI and Noob Dev
Role Looking for : Chairperson of the Board of Directors

Create a compelling cover letter that focuses on the strongest match between the candidate and this specific role.
Use only the information provided. Do not invent anything.
Return only the final cover letter.

"""
```

```python
openai = OpenAI(api_key=api_key)

response = openai.chat.completions.create(
    model="gpt-4o-mini",    
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
)

result = response.choices[0].message.content
```

```python
from convert import create_cover_letter_pdf

letter = create_cover_letter_pdf (
content=result, 
output_filename="cover_letter-pdf", 
applicant_name="Kavindu Sandaruwan",
contact_info="kavindu.sandaruwan@email.com | +94 71 234 5678 | Colombo, Sri Lanka",
date="August 23, 2026",

)
```
