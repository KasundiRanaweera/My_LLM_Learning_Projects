```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
```

```python
openai = OpenAI(api_key=api_key)
```

```python
messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Hi I'm Kasundi"}
    ]
```

```python
response = openai.chat.completions.create(
    model="gpt-4o-mini",   
    messages=messages
)

response.choices[0].message.content
```

Each api call is stateless (It's mean one api call things doesnt know other api call things - no memory)

```python
messages=[
        {"role": "user", "content": "What is my name?"}
    ]

response = openai.chat.completions.create(
    model="gpt-4o-mini",  
    messages=messages
)
response.choices[0].message.content

#This does not work because the model does not have memory of previous messages. 
#Each request is stateless and does not retain context from previous interactions.
#To maintain context, you would need to include the previous messages in the new request.
```
