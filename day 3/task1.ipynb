Using Open AI (Open API Key) connect to other APIs

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
load_dotenv(override=True)

openai_api_key = os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
groq_url = "https://api.groq.com/openai/v1"
openrouter_url = "https://api.openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"
```

```python
openai = OpenAI(api_key=openai_api_key)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)  
openrouter = OpenAI(api_key=openrouter_api_key, base_url=openrouter_url)
ollama = OpenAI(base_url=ollama_url, api_key="ollama")
#for ollama, there is no api key because it is running locally, 
# so we can use any string as the api key.
```

```python
tell_a_joke = [
    {"role": "user",
     "content": "Tell me a joke for a student on the topic of AI and LLMs."
    }]
```

```python
response = ollama.chat.completions.create(
    model="gemma3:270m",
    messages= tell_a_joke
)

result = response.choices[0].message.content
print(result)
```

```python
display(Markdown(result))
```

Make model more intelligent - Inference time scalling(use better prompts)

Reasoning Efforts: minimal <low <medium < high

Using reasoning efforts we can improve model performance.

```python
easy_puzzle =[
    {"role": "user","content": "You toss 2 coins. One of them is heads. Whats the probability the other is tails? Answer with the probabbility only."}
] #answer:2/3
```

```python
response = ollama.chat.completions.create(
    model="gemma3:270m", messages= easy_puzzle ,
    reasoning_effort="minimal")

result = response.choices[0].message.content
print(result)
```

```python
#using groq api to solve the easy puzzle
response = groq.chat.completions.create(
    model="openai/gpt-oss-120b", messages= easy_puzzle)

result = response.choices[0].message.content
display(Markdown(result))
```
