```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio
from IPython.display import display, Markdown
```

```python

load_dotenv(override=True)

openai_api_key = os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")

groq_url = "https://api.groq.com/openai/v1"
ollama_url = "http://localhost:11434/v1"

openai = OpenAI(api_key=openai_api_key)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
```

```python
system_prompt = "You are a helpful assistant"
```

craete a function and get user prompts

```python
def message_to_groq(user_prompt):
    response = groq.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "system", "content": system_prompt},
                  {"role": "user", "content": user_prompt}]
    )
    return response.choices[0].message.content
```

```python
message_to_groq("What is today's date?")
display(Markdown(message_to_groq("What is today's date?")))
```

```python
system_prompt = "You are a helpful assistant that responds in markdown without code blocks"


def call_gpt(user_prompt):
    response = groq.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "system", "content": system_prompt},
                  {"role": "user", "content": user_prompt}]
    )
    return response.choices[0].message.content
```

```python
import ollama


def call_ollama(user_prompt):
    response = ollama.chat.completions.create(
        model="gemma3:270m",
        messages=[{"role": "system", "content": system_prompt},
                  {"role": "user", "content": user_prompt}]
    )
    return response.choices[0].message.content
```

```python
def pick_model(prompt, model):
    if model == "GPT":
        result = call_gpt(prompt)

    elif model == "OLLAMA":
        result = call_ollama(prompt)

    else:
        raise ValueError("Unknown Error!")

    return result
```

```python
message_input = gradio.Textbox(label="Enter your messsage")
message_output = gradio.Markdown(label="Dia Response")
model_selector = gradio.Radio(choices=["GPT", "OLLAMA"], label="Select Model")

gradio.Interface(
    # call the created module
    fn=pick_model,
    inputs=[message_input, model_selector],
    outputs=message_output,
    title="Dia Dialog",
    description="A simple interface to interact with the GPT model",
    flagging_mode="never"

).launch()
```

# Building a company bruchure

```python
from scraper import fetch_website_contents
```

```python
system_message = """

You are an assistant that analyzes the contents of a company website landing page
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
"""
```

```python
def call_brochure_generator(company_name, url, model):

 company_data = fetch_website_contents(url)
 user_prompt = "Please create a short brochure about the company named " + company_name + " based on the following information: " + company_data

 if model == "GPT":
     result = call_gpt(user_prompt)
 elif model == "OLLAMA":
    result = call_ollama(user_prompt)
 else:
    raise ValueError( "Unknown model")

 return result 
```

```python
result =call_brochure_generator("Groq", "https://groq.com", "GPT")
print(result)
```

```python
message_input = gradio.Textbox(label="Enter your company name")
url_input = gradio.Textbox(label="Enter your company website URL")
message_output = gradio.Markdown(label="Response")
model_selector = gradio.Radio(choices=["GPT", "OLLAMA"], label="Select Model")

gradio.Interface(
    # call the created module
    fn=call_brochure_generator,
    inputs=[message_input, url_input, model_selector],
    outputs=message_output,
    title="Dia Dialog",
    description="A simple interface to interact with the GPT model",
    flagging_mode="never"

).launch()
```
