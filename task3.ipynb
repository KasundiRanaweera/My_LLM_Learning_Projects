```python
#print history
def chat(message, history):
    cleaned_history = []

    for h in history:
        cleaned_history.append({"role": h['role'], "content": h['content']})

    messages = [{"role": "system", "content": system_prompt}] + cleaned_history + [{"role": "user", "content": message}]
    response = groq.chat.completions.create(
        model= "openai/gpt-oss-120b",
        messages=messages
    )

    return response.choices[0].message.content

gradio.ChatInterface(fn=chat).launch()
#llm is statless so we have to pass the history to it every time we call it.
# The history is a list of dictionaries with the role and content of each message.
# We can use this to create a conversation with the model.
```

# DSI Chatbot


```python
##shot prompting
#single shot prompting: One-shot prompting involves providing a single example of the desired output format or style.

## multi-shot prompting: Multi-shot prompting involves providing multiple examples of the desired output format or style. 
# This can help the model better understand the context and produce more accurate responses.

system_prompt = """
You are a helpful assistant in a clothes store. You should try to gently encourage 
the customer to try items that are on sale. Hats are 60% off, and most other items are 50% off. 

For example, if the customer says "I'm looking to buy a hat', 
you could reply something like, "Wonderful - we have lots of hats - including several that are part of our sales event. "
Encourage the customer to buy hats if they are unsure what to get.
If the customer asks for shoes, you should respond that shoes are not on sale today, but you have a great selection of hats that are on sale. 
"""
```

```python
def chat(message, history):
    cleaned_history = []

    for h in history:
        cleaned_history.append({"role": h['role'], "content": h['content']})

    messages = [{"role": "system", "content": system_prompt}] + cleaned_history + [{"role": "user", "content": message}]
    response = groq.chat.completions.create(
        model= "openai/gpt-oss-120b",
        messages=messages
    )

    return response.choices[0].message.content

gradio.ChatInterface(fn=chat).launch()
```
